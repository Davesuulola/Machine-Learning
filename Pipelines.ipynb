<h3>A machine learning Pipeline is a way of putting all the steps your model needs into one organized sequence, so the data automatically passes through each step in the correct order.

Think of it like a production line in a factory:

Raw data
   →
Clean / scale data
   →
Train model
   →
Make prediction


Instead of manually doing each step every time, you put them together:

```python
"""Pipeline([
    ("scaler", StandardScaler()),
    ("model", KNeighborsClassifier())
])"""

##isntead of scaling first, then creating model, we combine those steps
```

Now you can essentially tell it:

```python
#pipeline.fit(X_train, y_train)
```

The pipleline handles 

X_train
   →
StandardScaler
   →
KNN

<h5>Pipelines work wonderfully well with cross validation. Each fold can perform its preprocessing separately, which helps prevent data leakage.

So for each model, we can build their own pipelines

<h3> Exercise A : Randomized Search using RF

```python
from sklearn.datasets import load_iris

iris_data = load_iris()
X = iris_data.data
y = iris_data.target
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import RandomizedSearchCV
```

```python
model = RandomForestClassifier(
    random_state = 42
)
```

```python
param_distributions = {
    "n_estimators": [10, 50, 100, 200, 300],
    "max_depth": [2, 3, 5, 10, None],
    "min_samples_split": [2, 5, 10]
}
```

```python
random = RandomizedSearchCV(
     model,
    param_distributions = param_distributions,
    n_iter = 10,
    cv =5,
    scoring = "accuracy",
    random_state = 42
)
```

```python
random.fit(X, y)
```

```python
print(f"Best combination: {random.best_params_}") 
print(f"\n Best Score: {random.best_score_}")
```

<h3>Exercise B: Pipleine + Grid search using KNN

```python
knn_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("knn_model", KNeighborsClassifier())
])
```

```python
param_grid = {
    "knn_model__n_neighbors" : [3,5,7,9,11]
}

#The double underscore points n_neighbors to the knn_model model
```

```python
#We then use Grid search

grid = GridSearchCV(
    knn_pipeline,
    cv = 5,
    param_grid = param_grid,
    scoring = "accuracy"
)
```

```python
grid.fit(X, y)
```

<h4>Notice how the pipeline is treated as an estimator. So in the pipeline, scaling and model creation happens

```python
print(f"Best combination: {grid.best_params_}") 
print(f"\n Best Score: {grid.best_score_}")
```
