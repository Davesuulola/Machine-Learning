```python
import pandas as pd

data = {
    "distance_km": [1, 2, 3, 4, 5],
    "population": [1000, 5000, 10000, 50000, 100000],
    "coverage": [1, 1, 0, 0, 0]
}

df = pd.DataFrame(data)

print(df)
```

```python
#We want to predict coverage. 1is good coverage, 0 is poor
```

```python
X = df[["distance_km", "population"]]
y = df["coverage"]
```

```python
#Import scaler
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
#Now, we scale the features

X_scaled = scaler.fit_transform(X)
```

```python
print(X_scaled)
```

But what happened? What did fit_transform do?
It is actually a 2 in 1 operation 
What fit did was make the algorithm understand the scale of the features
It determines the mean and standard deviation of the features and then 
Transform uses them to transform the values
