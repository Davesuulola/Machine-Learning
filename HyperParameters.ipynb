But where did these values come from, what are they?

<h3>These are called hyperparameters, not parameters

Parameters

These are values the model learns from the data.

For example, Linear Regression learns coefficients.


Hyperparameters

These are settings we choose before training

KNN:
n_neighbors

Random Forest:
n_estimators
max_depth
min_samples_split
min_samples_leaf

Decision Tree:
max_depth
min_samples_split

An approach to Hyperparameter tuning is Grid search, it involves trying a predefined set of hyperparameter combinations and see which performs best.

```python
#Let's test this with iris data
from sklearn.datasets import load_iris
import pandas as pd

iris_data = load_iris()
#print(iris_data)
```

```python
X = iris_data.data
y = iris_data.target
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV
```

```python
model = RandomForestClassifier(random_state = 42)
#No need to tune random state

#Let's apply grid search
```

```python
param_grid = {
    "n_estimators" : [10, 50, 100],
    "max_depth" : [2, 3, 5, None ]
}

grid_search = GridSearchCV(
    model,
    param_grid,    
    cv = 5,
    scoring = "accuracy"
)
#Notice i didnt add X and y, that's because we'll train the model after
```

```python
grid_search.fit(X, y)
```

```python
print("Best parameters:", grid_search.best_params_)
print("Best CV score:", grid_search.best_score_)
```

```python
#so we see that the best parameter for this RF model had a depth of 3, and 10 trees
```

<h2>There's a problem with Grid Search

Imagine using RF, and we want to use a parameter grid for the grid search, say RF
And we have 10 n_estimators and 10 max_depth. Grid search would have to evaluate 10 max_depth for each estimator, That's like 1000 combinations!

Pair that up with cross validation, say we have 5 folds, so it would perform 1000 cobimnation 5 times!, It would be computationally intensive

<h4>As the number of hyperparameters and possible values grows, the number of combinations can grow very quickly.

Imagine 5 hyperparameters
× 10 possible values each, thats 10 ** 5 =  100,000 combinations, with a 5 fold CV, 500000

<h3>That's where Randomized Search comes in

Think of a restaurant.

Grid Search

You say:

"I want to try every meal combination on this menu."

Even if there are 10,000 combinations.

Randomized Search

You say:

"Randomly try 50 combinations and find a good one."

You're not guaranteeing that you've tested every possibility, but you can search a large space much more efficiently.

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
#Let's use the same RF model
```

```python
param_distributions = {
    "n_estimators": [50, 100, 200, 300, 500],
    "max_depth": [3, 5, 10, 15, None],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4]
}
#look at all these hyperparameters, with Grid Search, it'd be an insane amount of combinations
```

```python
random_search = RandomizedSearchCV(
    model,
    param_distributions = param_distributions,
    n_iter = 20,
    cv =5,
    scoring = "accuracy",
    random_state = 42
    
)
```

The important parameter here is n_iter=20

It means:

Try 20 randomly selected hyperparameter combinations.

Instead of testing everything.

```python
random_search.fit(X, y)
```

```python
print(random_search.best_params_)
```
