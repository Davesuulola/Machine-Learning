<h2>What information am I actually giving the algorithm?

Feature engineering is basically:

Taking the information you already have and creating or transforming features so they provide more useful information to the model.

<h3>By carefully engineering the features, we provide the models better info to make better predictions

In the course of feature engineering, we can even provide better relationships between the features

Imagine the model sees:

Population = 30,000
Area = 30

It has to figure out the relationship itself.

But if we give it:

Population Density = 1,000 people/km²

we've already represented something that may be more directly related to network demand.

<h4>We're essentially saying:

"Here's a more useful way of representing the information."

This is particularly important in the fyp

<h3>Feature engineering isn't about creating as many features as possible. It's about creating meaningful representations of the underlying problem.

```python
import pandas as pd

data = {
    "population": [10000, 20000, 30000],
    "area_km2": [10, 10, 30]
}

df = pd.DataFrame(data)
df.head()
```

```python
#Let's create a new feature
df["population_density"] = (
    df["population"] / df["area_km2"]
)
df.head()
```

Consequently, feature engineering is about presenting a real world probelm in a way that makes it easier for the model to learn useful patterns.

<h3>When engineering a new feature, it mustn't be the target. 

The fearure must be independent of the target

Before adding any feature, ask:

1. Is it available at prediction time?
2. Is it independent of the target?
3. Does it have a reasonable physical/domain justification?
4. Does it actually improve validation performance?
