<h2>A model can get 100% on a tiny test set and still not be trustworthy.

Instead of just splitting, the data into say 80 and 20%. We split the data into K-folds.


<h3>K-Fold Cross validation

Say we have k = 5 , we split the data into 5 folds and run 5 rounds of training and testing
Round 1 : Train: Fold 2,3,4,5
Test:  Fold 1

Round 2 : Train: Fold 1,3,4,5
Test:  Fold 2
....
Round 5 : Train: Fold 1,2,3,4
Test:  Fold 5

The final test set should remain untouched until you're ready for the final evaluation.

Cross-validation usually happens inside the training data.

This is another way of protecting against accidentally tuning your model to the test set.

Cross-validation does not magically make a model better.

It gives you a better estimate of performance and helps you compare/tune models more reliably.

It also helps detect instability

```python
#Let's test it out
#Surprisingly, scikit learn has it's own data sets
from sklearn.datasets import load_iris
#The Iris dataset is a classic small classification dataset.
#Load it:
```

```python
import pandas as pd

iris_data = load_iris()
#Check it out
print(iris_data)
```

```python
#No need for this. data = pd.DataFrame(iris_data)
X = iris_data.data
y = iris_data.target
#print(X)
```

```python
#Create model. Note, cross validation is after the model is created, it automatically fits the model

from sklearn.ensemble import RandomForestClassifier

model  = RandomForestClassifier(
    n_estimators = 100,
    random_state = 42
)
```

```python
#Cross validate

from sklearn.model_selection import cross_val_score

scores = cross_val_score(
    model,
    X,
    y,
    cv = 5,
    scoring = "accuracy"
)
```

```python
print("Scores:", scores)
print("Mean:", scores.mean())
print("Standard deviation:", scores.std())
```
