<h1>What geographic area does this point belong to?

Imagine:

┌───────────────────────────┐

│         Area A            │

│      ● P1      ● P2       │

│                           │

└───────────────────────────┘


┌───────────────────────────┐

│         Area B            │

│          ● P3             │

│                           │

└───────────────────────────┘


A spatial join can tell us:

P1 → Area A
P2 → Area A
P3 → Area B

Then we can attach properties of that polygon to the point.

For example:

P1 → Area A → population = 12,000

P2 → Area A → population = 12,000

P3 → Area B → population = 8,500

```python
import geopandas as gpd
from shapely.geometry import Point, Polygon
```

```python
#Let's create  polygons, ie Areas
```

```python
areas = gpd.GeoDataFrame(
    {
        "area": ["Area_A", "Area_B"],
        "population": [12000, 8500]
    },
    geometry=[
        Polygon([
            (5.10, 7.27),
            (5.14, 7.27),
            (5.14, 7.31),
            (5.10, 7.31)
        ]),
        Polygon([
            (5.14, 7.27),
            (5.18, 7.27),
            (5.18, 7.31),
            (5.14, 7.31)
        ])
    ],
    crs="EPSG:4326"
)

#Normally, to create a GPD, we first create pd, then geometry column then combine, We just did all at once
#So sometimes if you want to create a GPD, you dont even need to create a PD df first
```

```python
points = gpd.GeoDataFrame(
    {
        "point": ["P1", "P2", "P3"],
    },
    geometry=[
        Point(5.12, 7.29),
        Point(5.13, 7.30),
        Point(5.16, 7.29)
    ],
    crs="EPSG:4326"
)
```

<h3>Perform the spatial join

```python
joined = gpd.sjoin(
    points,
    areas,
    how="left",
    predicate="within"
)
```

The important piece is:

predicate="within"

It means:

Find the polygon that each point is within.

```python
print(joined)
```

<h2>The spatial join has effectively attached polygon information to each point.
