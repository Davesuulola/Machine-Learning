How does a model decide what category/class a sample data belongs to?
A simple decision tree creates a tree by asking questions that divide data into classes

But we introduce the Logistic Regression model
This model creates probabilities of a sample, then using a threshold value, it then decides what class an observation belongs to


But whats the point? Isnt this just the same as a clasic DT or RF?
well, theLogistic regression introduces something important

<h3> Because of the probabilities, the model has stronger opinions on  different predictions

<h5>in fyp, if we have two locations and we're predicting coverage gap, one is 0.95 and the other 0.52 and threshold is 0.5, the first location is of a  stronger interest even thogh they are both gaps according to the threshold
    

```python
import pandas as pd

data = {
    "hours_studied": [1, 2, 3, 4, 5, 6, 7, 8],
    "passed": [0, 0, 0, 1, 1, 1, 1, 1]
}

df = pd.DataFrame(data)

X = df[["hours_studied"]]
y = df["passed"]
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()

model.fit(X, y)
```

```python
predictions = model.predict([[2], [3.5], [5], [7]])

print(predictions)
```

```python
#See that we still get the same results, but something else happened behind the curtains.
#to see what happened, lets see the probabilities the model predicted
```

```python
probabilities = model.predict_proba([[2], [3.5], [5], [7]])

print(probabilities)
```

```python
#What this means is that say for 2, prob of class 0 (0 = fail) = 0.84640881, prob of class 1 (pass = 0.15359119). adding them gives 1
#So the model is basically saying, for 2, there's a 84.64% chance it belongs to class 0 and a 15.59% chance it belongs to class 1
```
