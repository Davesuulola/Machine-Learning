```python
#Let's practice with the exam score exercise
#Step 1: define data

import pandas as pd

data = {
    "hours_studied" : [1, 2, 3, 4, 5, 6, 7, 8],
    "exam_score" : [42, 48, 55, 61, 68, 74, 81, 88]
}

df = pd.DataFrame(data)
```

```python
#Step 2: Define features and target
X = df[["hours_studied"]]
y = df["exam_score"]
```

```python
#Step 3: Split data
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size = 0.25,
    random_state = 42
)
```

```python
#Step 3: Create model
from sklearn.linear_model import LinearRegression

model = LinearRegression()
```

```python
#Step 4: Train/fit model

#note that you have to use X_train and y_train
model.fit(X_train, y_train)
```

```python
#Step 5: Make prediction
#Since we split the data, we dont have to input data. As in, we already set aside some data for testing
#model.predict(X_test)
#since we have many data in X_test, we say

predictions = model.predict(X_test)
#so this makes predictions on the values in X-test and gives us results, WHICH WE COMPARE TO y_test TO DETERMINE ACCURACY
```

```python
#let's see how accurate the model is
print(y_test)
print()
print(predictions)
```
