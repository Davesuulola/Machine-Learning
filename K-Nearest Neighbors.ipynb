<h2>KNN provides a different approach to modelling


It makes predictions on data by looking at the samples who have close or nearby values to the data that prediction is to be performed on

For example, we want to predict if some one who studied for 6.5 hours will pass, if 1,2,3,4 failed and 5,6,7 passed. 

The model look at the closest values to 6.5: 5,6,7. Since they passed, then 6.5 should as well

<h3>But how many nearby values does the model look at?

Thats what the K in KNN stands for. 
K = number of neighbors we look at.

So in the last example, k = 3

<h3> When the model looks at the neighbors, it looks at the majority and makes the decision

```python
import pandas as pd

data = {
    "hours_studied": [1, 2, 3, 4, 5, 6, 7, 8],
    "passed": [0, 0, 0, 1, 1, 1, 1, 1]
}

df = pd.DataFrame(data)

X = df[["hours_studied"]]
y = df["passed"]

df.head()
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
model = KNeighborsClassifier(n_neighbors = 3)

#Just like RF, KNN takes in a parameter to determine the no of neighbors
```

```python
model.fit(X,y)
```

```python
predictions = model.predict([[2], [3.5], [5], [7]])

print(predictions)
```

```python
#So for each of those values, KNN viewed the 3 closest samples to make a prediction
```

It doesn't really build a complicated mathematical model in the same sense as linear regression, DT and RF models.

It essentially stores the training observations and uses them when making a prediction.

That's why KNN is often described as a lazy learner.
