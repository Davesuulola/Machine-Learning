<h1>How to Choose an ML Algorithm

One of the biggest beginner mistakes is:

"Random Forest is good, so I'll use Random Forest."

That's not really ML reasoning.

<h3>A better process is:

Problem
   
What am I predicting?
   
What does the data look like?
   
What algorithms are suitable?
   
What preprocessing is needed?
   
How should I evaluate them?
   
Compare candidates

There is rarely a single algorithm that is universally best.

1. Problemm type. Is it a continuous number or classification problem? 

2. Data. How complicated is the data, are the relationships complicated?

3. Data size. How big is the data?

4. Interpretability. Some models are easier to explain than others, especially when the problem requires explanations on results
