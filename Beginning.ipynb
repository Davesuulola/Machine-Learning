```python
#let's create an ML model

import pandas as pd

#we form a dataset using a dictionary and covert using pandas to a DF
data = {
    "distance_km": [0.5, 1.0, 1.5, 2.0, 2.5, 3.0, 3.5],
    "signal_dbm": [-62, -68, -73, -79, -84, -89, -94]
}

df = pd.DataFrame(data)

print(df)
```

```python
#let's define the features and target. Recall that X is features and y is target or result
#using column selection, distance_km is input while signal_db is output
X = df[["distance_km"]]
y = df["signal_dbm"]
print(X)
print(y)
```

```python
#let use now train a model. Since we have a simple table with two values. we can see that as one value changes, another changes
# intuitively, we can use a straight line graph to imagine the table

#Therefore we use Linear Regression
from sklearn.linear_model import LinearRegression

model = LinearRegression()

# No we want the program to learn the relationship btw X and y, So we use the .fit() function
#"Here are my examples. Learn the relationship between X and y."
model.fit(X,y)
```

```python
#Now that it has learned the relationship, let us make a prediction
#To do that we can use the .predict(X_value) function
#Since X is a 2D structure, we still use 2 brackets
prediction = model.predict([[2.7]])

print(prediction)
```

```python
#Because its a linear regression model, it used the y =mx + c formula
# we can even check to see what our slope and intercept is
#since the signal decreases as distance increases, intuitively, the slope should be -ve
print("The slope is: " , model.coef_)
print("Intercept: " ,model.intercept_)
```
